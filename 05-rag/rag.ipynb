```python
!pip install -q sentence-transformers faiss-cpu transformers

from sentence_transformers import SentenceTransformer
import faiss, numpy as np
from transformers import pipeline

# 1) Our little document library (normally these are your PDFs or web pages).
docs = [
    "The CS department office is in Room 204 and opens at 9 AM.",
    "Dr. Example teaches Databases on Mondays at 10 AM.",
    "The campus library closes at 8 PM on weekdays.",
]

# 2) Turn each document into an embedding (meaning-as-numbers).
embedder = SentenceTransformer("all-MiniLM-L6-v2")
doc_vecs = np.array(embedder.encode(docs))

# 3) Store them in a FAISS index we can search by meaning.
index = faiss.IndexFlatL2(doc_vecs.shape[1])
index.add(doc_vecs)

# 4) A question -> embed it -> find the single closest document.
question = "When does Dr. Example teach Databases?"
q_vec = np.array(embedder.encode([question]))
_, hits = index.search(q_vec, k=1)
retrieved = docs[hits[0][0]]
print("Retrieved context:", retrieved)

# 5) Ask a small model, giving it ONLY the retrieved context.
gen = pipeline("text-generation", model="TinyLlama/TinyLlama-1.1B-Chat-v1.0",
               max_new_tokens=60)
prompt = f"Context: {retrieved}\nQuestion: {question}\nAnswer:"
print(gen(prompt)[0]["generated_text"])
```
